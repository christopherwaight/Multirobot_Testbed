# Vector Fields

## Uniform Flow

```python
import numpy as np
import matplotlib.pyplot as plt

# Create Meshgrid
x = np.arange(0, 10, 0.5)
y = np.arange(0, 10, 0.5)
x, y = np.meshgrid(x, y)

# X Uniform Flow
u = np.ones_like(x)  
v = np.zeros_like(y)  

fig, axes = plt.subplots(1,2, figsize=(8,4))

# Plot X Uniform Flow
axes[0].quiver(x, y, u, v)
axes[0].set_title("Uniform flow in x-direction")

# Y Uniform Flow
u = np.zeros_like(x)  # x-directed flow (zero)
v = np.ones_like(y)  # y-directed flow (uniform)

# Plot Y Uniform Flow
axes[1].quiver(x, y, u, v)
axes[1].set_title("Uniform flow in y-direction")

# Show Figure
plt.show()
```

## Source and Sink

```python
# Create Meshgrid
x = np.arange(0, 10, 0.5)
y = np.arange(0, 10, 0.5)
x, y = np.meshgrid(x, y)

# Source Location
x_source = 5
y_source = 5

# Centring over source
x_centre = x - x_source
y_centre = y - y_source

# Radial Distance from Centre
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate the source vectors at each point
u = x_centre / r**2
v = y_centre / r**2

u1 = -u
v1 = -v


# Quiver Plot 
#plt.figure(figsize=(8, 8))
#plt.quiver(x, y, u, v)
#plt.title("Source at (5, 5)")
#plt.show()

fig, axes = plt.subplots(1,2, figsize=(8,4))

# Plot X Uniform Flow
axes[0].quiver(x, y, u, v)
axes[0].set_title("Source")


# Plot Y Uniform Flow
axes[1].quiver(x, y, u1, v1)
axes[1].set_title("Sink")

# Show Figure
plt.show()
```

```python
# Calculating A Sink

# Sink Location
x_sink = 5
y_sink = 5

# Centring over sink
x_centre = x - x_sink
y_centre = y - y_sink

# Calculate the distance r from the center
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate the vector at ea. pt.
u = -x_centre / r**2
v = -y_centre / r**2

# Plot
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Sink at (5, 5)")
plt.show()
```

## Free and Fixed Vortices

```python
# Calculating a Vortex

# Calculate the angular component of the velocity
v_theta = 1 / r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Free Vortex at (5, 5)")
plt.show()
```

```python
# Fixed Vortex

# Calculate the angular component of the velocity
v_theta = 1 * r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Fixed Vortex at (5, 5)")
plt.show()
```

## Saddles

```python
# Calculating a Saddle

# Calculate the angular component of the velocity
v_theta = 1 / r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Saddle at (5, 5)")
plt.show()
```

## Dipole or Doublet

```python
# Dipole - Doublet

# Create Meshgrid
x = np.arange(0, 10, 0.4)
y = np.arange(0, 10, 0.4)
x, y = np.meshgrid(x, y)

x_centre1 = x - 5
y_centre1 = y - 5

x_centre2 = x - 5.01
y_centre2 = y - 5.01

# Calculate the distance r from the center
r1 = np.sqrt(x_centre1**2 + y_centre1**2)
r2 = np.sqrt(x_centre2**2 + y_centre2**2)

# Avoid division by zero at the center
r1[r1==0.000] = 1e-6
r2[r2==0.000] = 1e-6

#r1[r1<0.1] = 10
#r2[r2<0.1] = 10

# Calculate the source vector components
u = -x_centre1 / r1**2 + x_centre2 / r2**2
v = -y_centre1 / r1**2 + y_centre2 / r2**2

# Normalize the vectors
magnitude = np.sqrt(u**2 + v**2)
u /= magnitude
v /= magnitude

# Plot the Source
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Dipole at (5, 5)")
plt.show()
```

```python
# Dipole - Doublet

# Create Meshgrid
x = np.arange(0, 10, 0.2)
y = np.arange(0, 10, 0.2)
x, y = np.meshgrid(x, y)

x_centre1 = x - 5
y_centre1 = y - 5

x_centre2 = x - 5.1
y_centre2 = y - 5.1

# Calculate the distance r from the center
r1 = np.sqrt(x_centre1**2 + y_centre1**2)
r2 = np.sqrt(x_centre2**2 + y_centre2**2)

# Avoid division by zero at the center
r1[r1==0.000] = 1e-6
r2[r2==0.000] = 1e-6

r1[r1<0.2] = 100
r2[r2<0.2] = 100

# Calculate the source vector components
u = -x_centre1 / r1**2 + x_centre2 / r2**2
v = -y_centre1 / r1**2 + y_centre2 / r2**2


# Plot the Source
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Dipole at (5, 5)")
plt.show()
```

## Combining Different Fields

### Sink + Vortex

```python
# Calculating A Sink

# Create Meshgrid
x = np.arange(0, 10, 0.2)
y = np.arange(0, 10, 0.2)
x, y = np.meshgrid(x, y)

x_centre = x - 5
y_centre = y - 5

# Calculate the distance r from the center
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r<0.5] = 100

# Calculate the angular component of the velocity
v_theta = 1 / r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Calculate the sink vector components
u =  u - x_centre / r**2
v =  v - y_centre / r**2

# Plot the Source
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("sink at (2, 5)")
plt.show()
```

### Source + Uniform

```python
# Create Meshgrid
x = np.arange(0, 6, 0.25)
y = np.arange(0, 6, 0.25)
x, y = np.meshgrid(x, y)

# X Uniform Flow
u = np.ones_like(x)  
v = np.zeros_like(x)  


# Calculating A Source

x_centre = x - 2
y_centre = y - 3

# Calculate the distance r from the center
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate the source vector components
u = u + x_centre / r**2
v = v + y_centre / r**2

# Plot the Source
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)
plt.title("Source at (2, 5)")
plt.show()

# Plot downstream
# Define a mask to only consider x values greater than 3
 

# Normalize the vectors
magnitude = np.sqrt(u**2 + v**2)
#u /= magnitude
#v /= magnitude

mask = (x > 4.95) & (x < 5.05)

# Plot for x > 3
plt.figure(figsize=(8, 8))
plt.quiver(x[mask], y[mask], u[mask], v[mask])
plt.title("Downstream ~~ 4.8")
plt.show()

plt.plot( magnitude[mask].flatten(),y[mask].flatten(),)
plt.xlabel('Magnitude')
plt.ylabel('Y Value')
plt.title("Magnitude at different Y values")
plt.show
```

### Rankine Vortex

```python
import numpy as np
import matplotlib.pyplot as plt

R = 1

# Create Meshgrid
x = np.arange(0, 6, 0.22)
y = np.arange(0, 6, 0.22)
x, y = np.meshgrid(x, y)

# Source Location
x_source = 3
y_source = 3

# Centring over source
x_centre = x - x_source
y_centre = y - y_source

# Radial Distance from Centre
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6


# Calculate the angular component of the velocity
v_theta = 1 / r 
v_theta[r < R] = r[r<R]

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(6, 6))
plt.quiver(x, y, u, v)
plt.title("Rankine Vortex (3, 3)")
plt.show()
```

```python
# Corner Flow

# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Define alpha for 90 degree corner
alpha = 2

# Create Meshgrid
x = np.arange(-0.1, 10, 0.2)
y = np.arange(-0.1, 10, 0.2)
X, Y = np.meshgrid(x, y)

# Calculate r, theta
r = np.sqrt(X**2 + Y**2)
theta = np.arctan2(Y, X)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate velocity components in polar coordinates
u_r = alpha * r**(alpha - 1) * np.cos(alpha * theta)
u_theta = -alpha * r**(alpha - 1) * np.sin(alpha * theta)

# Convert to cartesian coordinates
U = u_r * np.cos(theta) - u_theta * np.sin(theta)
V = u_r * np.sin(theta) + u_theta * np.cos(theta)

# Create the figure and the subplots
fig, ax = plt.subplots(figsize=(6, 6))

# Plot the corner flow as streamlines
ax.streamplot(X, Y, U, V, density=2, arrowstyle='->', arrowsize=1.5)

# Set aspect ratio to be equal
ax.set_aspect('equal')

ax.set_title("90 Degree Corner Flow")

# Show the plots
plt.tight_layout()
plt.show()
```

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Define alpha for 90 degree corner
alpha = 1.5 

# Create Meshgrid
x = np.arange(-10, 10, 0.5)
y = np.arange(-0.01, 10, 0.5)
X, Y = np.meshgrid(x, y)

# Calculate r, theta
r = np.sqrt(X**2 + Y**2)
theta = np.arctan2(Y, X)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate velocity components in polar coordinates
u_r = alpha * r**(alpha - 1) * np.cos(alpha * theta)
u_theta = -alpha * r**(alpha - 1) * np.sin(alpha * theta)

# Convert to cartesian coordinates
U = u_r * np.cos(theta) - u_theta * np.sin(theta)
V = u_r * np.sin(theta) + u_theta * np.cos(theta)
U = -U
V = -V

# Normalize the velocity components for better visualization
magnitude = np.sqrt(U**2 + V**2)
U /= magnitude
V /= magnitude

# Create the figure and the subplots
fig, ax = plt.subplots(figsize=(6, 6))

# Plot the corner flow as quiver
ax.quiver(X, Y, U, V, scale=20)

# Set aspect ratio to be equal
ax.set_aspect('equal')

ax.set_title("120 Degree Corner Flow")

# Show the plots
plt.tight_layout()
plt.show()
```
