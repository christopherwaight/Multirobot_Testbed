```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create coordinate grids
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

# Create a scalar field with two hills
Z = 2 * np.exp(-(X-1)**2 - (Y-1)**2) + 1.5 * np.exp(-((X+1)**2 + (Y+1)**2)/2) + 0.1*np.sin(X*Y)

# Create figure with 3 subplots
fig = plt.figure(figsize=(15, 5))

# 1. 3D Surface Plot
ax1 = fig.add_subplot(131, projection='3d')
surf = ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.set_title('3D Scalar Field')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Field Value')

# 2. Contour Plot
ax2 = fig.add_subplot(132)
contour = ax2.contour(X, Y, Z, levels=15, colors='black', linewidths=0.5)
contourf = ax2.contourf(X, Y, Z, levels=15, cmap='viridis', alpha=0.7)
ax2.clabel(contour, inline=True, fontsize=8)
ax2.set_title('Contour Map')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_aspect('equal')

# 3. Grayscale Heatmap
ax3 = fig.add_subplot(133)
heatmap = ax3.imshow(Z, extent=[-3, 3, -3, 3], origin='lower', cmap='gray', aspect='equal')
ax3.set_title('Grayscale Heatmap')
ax3.set_xlabel('X')
ax3.set_ylabel('Y')

# Add colorbars
fig.colorbar(surf, ax=ax1, shrink=0.5)
fig.colorbar(contourf, ax=ax2)
fig.colorbar(heatmap, ax=ax3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- 1.  Define the vortex ----------
n = 200
x = np.linspace(-1, 1, n)
y = np.linspace(-1, 1, n)
X, Y = np.meshgrid(x, y)

U = -Y                 # u = -y
V =  X                 # v =  x
speed = np.hypot(U, V) # magnitude field

# ---------- 2.  Image 1 – contour *lines* ----------
plt.figure(figsize=(5, 5))
levels = np.linspace(speed.min(), speed.max(), 20)  # 20 line levels
cs = plt.contour(X, Y, speed, levels=levels, colors='k', linewidths=0.8)
plt.clabel(cs, fmt='%.2f', fontsize=6)              # optional labels on the lines
plt.title('Vortex – speed contour lines')
plt.axis('equal')
plt.tight_layout()

# ---------- 3.  Image 2 – greyscale map ----------
plt.figure(figsize=(5, 5))
plt.imshow(speed, extent=[-1, 1, -1, 1], origin='lower', cmap='gray')
plt.title('Vortex – speed (greyscale)')
plt.axis('equal')
plt.tight_layout()

plt.show()
```
