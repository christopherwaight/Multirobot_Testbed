**One Point Simulated Annealing Example**

```python
#Finding the Maximum

import numpy as np
import matplotlib.pyplot as plt
import math

# Define a 1D function to minimize
def func(x):
    #return -x**2 + 10*x + 25   # This is a simple parabola
    return -1*x**2 + 10*x + 25 + 40*np.sin(0.21*np.pi*x)   
    
# Simulated annealing algorithm
def simulated_annealing(func, initial_state, initial_temp, cooling_rate, num_steps):
    current_state = initial_state
    current_temp = initial_temp

    states = []
    temperatures = []

    for step in range(num_steps):
        neighbor_state = current_state + np.random.uniform(-2, 2)  # Generate a new state
        cost_diff = func(neighbor_state) - func(current_state)  # Compute  difference

        # If the new state is better, accept it
        if cost_diff > 0:
            current_state = neighbor_state
        # If the new state is worse, accept it with a probability of e^(-cost_diff/current_temp)
        elif np.random.rand() < math.exp(cost_diff / current_temp): #-cost for minimization function
            current_state = neighbor_state

        # Cool down
        current_temp *= cooling_rate

        states.append(current_state)
        temperatures.append(current_temp)

    return states, temperatures

# Set initial parameters
initial_state = 20
initial_temp = 100
cooling_rate = 0.99
num_steps = 200

# Run the simulated annealing algorithm
states, temperatures = simulated_annealing(func, initial_state, initial_temp, cooling_rate, num_steps)

# Plot the states over time
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.plot(states)
plt.title('State over time')
plt.xlabel('Time step')
plt.ylabel('State')

# Plot the temperature over time
plt.subplot(1,2,2)
plt.plot(temperatures)
plt.title('Temperature over time')
plt.xlabel('Time step')
plt.ylabel('Temperature')
plt.tight_layout()
plt.show()

#Plotting the Function over the space
x_values = np.linspace(-10, 25, 400)
y_values = func(x_values)
plt.figure()
plt.plot(x_values, y_values, label='f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Landscape being traversed')
plt.legend()
plt.show()

max_index = np.argmax(y_values)
x_max = x_values[max_index]
print("Index of max value:", x_max)
```

Simulated Annearling In 2 Dimensions - 1 Point

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Define a 2D function to minimize
def func(x, y):
    return -1*(x**2 + y**2) + 2*(x + y) + 25 + 30*np.sin(0.2*np.pi*(x+y))

# Simulated annealing algorithm
def simulated_annealing(func, initial_state, initial_temp, cooling_rate, num_steps):
    current_state = initial_state
    current_temp = initial_temp

    states = []
    temperatures = []

    for step in range(num_steps):
        neighbor_state = current_state + np.random.uniform(-2, 2, 2)  # Generate a new state
        cost_diff = func(*neighbor_state) - func(*current_state)  # Compute  difference

        # If the new state is better, accept it
        if cost_diff > 0:
            current_state = neighbor_state
        # If the new state is worse, accept it with a probability of e^(-cost_diff/current_temp)
        elif np.random.rand() < math.exp(cost_diff / current_temp):
            current_state = neighbor_state

        # Cool down
        current_temp *= cooling_rate

        states.append(current_state)
        temperatures.append(current_temp)

    return states, temperatures

# Set initial parameters
initial_state = np.array([-20, -20])
initial_temp = 80
cooling_rate = 0.99
num_steps = 80

# Run the simulated annealing algorithm
states, temperatures = simulated_annealing(func, initial_state, initial_temp, cooling_rate, num_steps)

# Create a mesh grid
x = np.linspace(-20, 20, 100)
y = np.linspace(-20, 20, 100)
X, Y = np.meshgrid(x, y)
Z = func(X, Y)

# Create a 2D plot with contours
plt.figure(figsize=(8, 8))
plt.contourf(X, Y, Z, 50, cmap='viridis')
plt.colorbar(label='f(X, Y)')
plt.title('Simulated Annealing Path')

# Plot the path
states = np.array(states)
plt.plot(states[:, 0], states[:, 1], 'r-', label='Path')

plt.xlabel('X')
plt.ylabel('Y')

plt.legend()
plt.show()


# Create a 3D plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.5)

# Plot the path
states = np.array(states)
ax.plot3D(states[:, 0], states[:, 1], func(states[:, 0], states[:, 1]), 'r-', label='Path')
ax.scatter3D(states[:, 0], states[:, 1], func(states[:, 0], states[:, 1]), c='r')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('f(X, Y)')
ax.set_title('Simulated Annealing Path')
```
