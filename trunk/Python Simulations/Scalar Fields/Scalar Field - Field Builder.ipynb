# Using Sines and Cosines

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create Meshgrid
x = np.arange(0, 15, 0.1)
y = np.arange(0, 15, 0.1)
X,Y = np.meshgrid(x,y)
Z = 2*np.sin(X) + 4*np.sin(0.8*Y)


fig = plt.figure(figsize=(16,8))  # Adjusted the figure size to accommodate both plots

# Plot Topo-Style Map
ax1 = fig.add_subplot(1, 2, 1)
ax1.imshow(Z, cmap='viridis', origin='lower', extent=[0, 10, 0, 10])
ax1.set_title("Landscape of Sine Waves")

# Plot 3D surface
ax2 = fig.add_subplot(1, 2, 2, projection='3d') 
ax2.plot_surface(X, Y, Z, cmap='viridis')
ax2.set_title("Surface Map of Sine Waves")

plt.show()
```

## Gaussians

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define a Gaussian RBF
def gaussian_rbf(x, y, x0, y0, amplitude=1, sigma_x=1, sigma_y=1):
    return amplitude * np.exp(-((x - x0)**2 / (2 * sigma_x**2) + (y - y0)**2 / (2 * sigma_y**2)))

# Create a grid of points
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
x, y = np.meshgrid(x, y)

# The centers of the Gaussians
centers = [(0, y) for y in np.linspace(-7.5, 7.5, 10)]

# The amplitudes of the Gaussians
amplitudes = np.linspace(0.5, 2, len(centers))

# Create the ridge by superimposing the Gaussians
z = np.zeros_like(x)
for center, amplitude in zip(centers, amplitudes):
    z += gaussian_rbf(x, y, center[0], center[1], amplitude, sigma_x=2, sigma_y=2)

# Plot the resulting shape
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')
plt.show()
```

```python
import numpy as np
from scipy.special import jn, jn_zeros  # Bessel function of the first kind and its zeros
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def drumhead(n, k, distance, angle, t):
    nth_zero =jn_zeros(n, k)[-1]  # Using the last (kth) zero
    return np.cos(t)*np.cos(n*angle)*jn(n, distance*nth_zero)

# Create a polar grid of coordinates
theta = np.r_[0:2*np.pi:50j]
radius = np.r_[0:1:50j]

x = np.array([r*np.cos(theta) for r in radius])
y = np.array([r*np.sin(theta) for r in radius])
z = np.array([drumhead(1, 1, r, theta, 0.5) for r in radius])

# Create a 3D plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, rstride=1, cstride=1, cmap='viridis', alpha=0.6, linewidth=0)

# Add some additional settings for the plot
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Drumhead')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the function
def func(x, y):
    return 0.05-0.001*(x**2 + y**2)

# Generate x, y values
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)
x, y = np.meshgrid(x, y)

# Compute z values
z = func(x, y)

# Create the figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Create the plot
_x = x.flatten()
_y = y.flatten()
_z = z.flatten()
dx = dy = 1 # width and depth of the bars
dz = _z # heights of the bars
ax.bar3d(_x, _y, np.zeros_like(dz), dx, dy, dz, color='green')

# Manually setting the z-axis limits
ax.set_zlim(-0.03, 0.05)

plt.title("Discrete 3D surface plot")
plt.show()

```
