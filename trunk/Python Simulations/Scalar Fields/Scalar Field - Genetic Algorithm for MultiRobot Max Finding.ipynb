```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


def f(x, y):
    return -0.59 * x ** 2 + 1.5 * x - 0.75 * y ** 2 + 2.5 * y + 15*np.sin(.8*x) + 40*np.sin(0.5*y) + 100 -0.4*x*y  + 0.5*np.sin(0.8*x*y + 2.2*y/x + (x+y)**2) - .2*np.sin(0.7*x*x/y + 42*y/x + (x-y)**2)

def initialize_population(pop_size):
    return np.random.uniform(-12, -8, (pop_size, 2))

def select_parents(population, fitness):
    idx = np.argsort(fitness)[-2:]  # Select the top 2 individuals... threesomes not allowed
    return population[idx]

def crossover(parents):
    np.random.shuffle(parents)
    children = []
    for i in range(0, len(parents), 2):
        child1 = [parents[i][0], parents[i+1][1]]
        child2 = [parents[i+1][0], parents[i][1]]
        children.extend([child1, child2])
    return np.array(children)

def mutate(children):
    mutation_chance = 0.3
    mutation_scale = 2
    for child in children:
        if np.random.rand() < mutation_chance:
            child += np.random.normal(0, mutation_scale, 2)
    return children

def genetic_algorithm():
    pop_size = 3
    generations = 101
    colors = ['red', 'yellow', 'blue']

    population = initialize_population(pop_size)
    for generation in range(generations):
        fitness = [f(x, y) for x, y in population]
        parents = select_parents(population, fitness)
        children = crossover(parents)
        children = mutate(children)
        population[np.argsort(fitness)[:len(children)]] = children

        if generation % 10 == 0:
            print(f"Generation {generation}: Current fitness {(fitness)}")

    fitness = [f(x, y) for x, y in population]
    best_solution = population[np.argmax(fitness)]
    print(f"Found maximum at x={best_solution[0]}, y={best_solution[1]}")
    print(f"Max value: {f(*best_solution)}")

genetic_algorithm()
```

```python
# This method requires memory. To execute in a reactive way, you would need more robots.

def differential_evolution():
    pop_size = 4
    generations = 101
    F = 1.8 # Scaling factor
    CR = 0.7 # Crossover probability

    population = initialize_population(pop_size)
    
    for generation in range(generations):
        for i in range(pop_size):
            target = population[i]
            idxs = [idx for idx in range(pop_size) if idx != i]
            a, b, c = population[np.random.choice(idxs, 3, replace=False)]
            
            # Mutation
            mutant = a + F * (b - c)

            # Crossover
            trial = np.where(np.random.rand(2) < CR, mutant, target)

            # Selection
            if f(*trial) > f(*target):
                population[i] = trial

        if generation % 10 == 0:
            fitness = [f(x, y) for x, y in population]
            print(f"Generation {generation}: Current fitness {(fitness)}")
            
    best_solution = population[np.argmax([f(x, y) for x, y in population])]
    print(f"Found maximum at x={best_solution[0]}, y={best_solution[1]}")
    print(f"Max value: {f(*best_solution)}")
    
differential_evolution()
```
