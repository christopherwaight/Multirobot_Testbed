# Vector Fields - Regions of Interest

## 1. Centre of A Sink/ Source and Constant Radius

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Create Meshgrid
x = np.arange(0, 10, 0.5)
y = np.arange(0, 10, 0.5)
X, Y = np.meshgrid(x, y)

# Source Location
x_source = 5
y_source = 5

# Centring over source
x_centre = X - x_source
y_centre = Y - y_source

# Radial Distance from Centre
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate the source vectors at each point
u = x_centre / r**2
v = y_centre / r**2

plt.figure(figsize=(8, 8))

# Quiver plot
plt.quiver(X, Y, u, v)

# Add source point
plt.scatter(x_source, y_source, color='red', marker='x', linewidths=2.5)

# Create circle
circle = plt.Circle((5,5), 4, color='red', fill=False)

# Add the circle to the plot
plt.gca().add_patch(circle)

# Set aspect ratio to be equal, so the circle looks circular
plt.gca().set_aspect('equal')

plt.title("Source at (5, 5) with Circle")
plt.show()
```

## 2. Centre and constant radius of a Free vortex

```python
# Calculating a Vortex

# Calculate the angular component of the velocity
v_theta = 1 / r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(8, 8))
plt.quiver(X, Y, u, v)
plt.title("Free Vortex at (5, 5)")

# Add source point
plt.scatter(x_source, y_source, color='red', marker='x', linewidths=2.5)

# Create circle
circle = plt.Circle((5,5), 4, color='red', fill=False)

# Add the circle to the plot
plt.gca().add_patch(circle)

# Set aspect ratio to be equal, so the circle looks circular
plt.gca().set_aspect('equal')

plt.show()
```

## 5. Sink and Source and Bifurcation lines of a Dipole

```python
# Dipole - Doublet

# Create Meshgrid
x = np.arange(0, 10, 0.4)
y = np.arange(0, 10, 0.4)
x, y = np.meshgrid(x, y)

x_centre1 = x - 4.75
y_centre1 = y - 4.75

x_centre2 = x - 5.25
y_centre2 = y - 5.25

# Calculate the distance r from the center
r1 = np.sqrt(x_centre1**2 + y_centre1**2)
r2 = np.sqrt(x_centre2**2 + y_centre2**2)

# Avoid division by zero at the center
r1[r1==0.000] = 1e-6
r2[r2==0.000] = 1e-6


# Calculate the source vector components
u = -x_centre1 / r1**2 + x_centre2 / r2**2
v = -y_centre1 / r1**2 + y_centre2 / r2**2

#Calculating Magnitude at each point
magnitude = np.sqrt(u**2 + v**2)


# Masking
mask = np.logical_and(magnitude >= 0.05, magnitude <= 0.07)
u_masked = np.where(mask, u, 0)
v_masked = np.where(mask, v, 0)

# Normalize the masked vectors
magnitude_masked = np.sqrt(u_masked**2 + v_masked**2)
u_masked /= magnitude_masked
v_masked /= magnitude_masked


# Normalize the vectors
u /= magnitude
v /= magnitude

# Plot the Dipole
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v, alpha = 1)

# Plot the Mask
plt.quiver(x, y, u_masked, v_masked, color='blue', alpha = 1)

# Draw contour lines where the magnitude is between 0.05 and 0.07
contour_levels = [0.8, 1]
plt.contour(x, y, magnitude, contour_levels, colors='blue')

# Add source point
plt.scatter(4.75, 4.75, color='red', marker='x', linewidths=2.5)

# Add source point
plt.scatter(5.25, 5.25, color='orange', marker='o', linewidths=2.5)

# Adding Bifurcation line
bifn = np.arange(5.25, 10, 0.5)
plt.plot(bifn,bifn,'--','blue')

# Title and Reveal
plt.title("Dipole Source, Sink, Equipotential Line and Bifurcation line")
plt.show()
```

## 6. Centre and radius of a Sink+Vortex

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Create Meshgrid
x = np.arange(0, 10, 0.2)
y = np.arange(0, 10, 0.2)
x, y = np.meshgrid(x, y)

x_source = 5
y_source = 5

x_centre = x - x_source
y_centre = y - y_source

# Calculate the distance r from the center
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r<0.5] = 100

# Calculate the angular component of the velocity
v_theta = 1 / r 

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Calculate the sink vector components
u =  u - x_centre / r**2
v =  v - y_centre / r**2




# Plot the Sink and Vortex
plt.figure(figsize=(8, 8))
plt.quiver(x, y, u, v)

# Create circle
circle = plt.Circle((5,5), 2, color='red', fill=False)

# Add the circle to the plot
plt.gca().add_patch(circle)

# Set aspect ratio to be equal, so the circle looks circular
plt.gca().set_aspect('equal')


# Add source point
plt.scatter(x_source, y_source, color='red', marker='x', linewidths=2.5)

plt.title("sink at (5, 5 and radius of constant magnitude)")
plt.show()
```

## 7. Centre of a Rankine Vortex w/ Eye

```python
import numpy as np
import matplotlib.pyplot as plt

R = 1

# Create Meshgrid
x = np.arange(0, 6, 0.22)
y = np.arange(0, 6, 0.22)
x, y = np.meshgrid(x, y)

# Source Location
x_source = 3
y_source = 3

# Centring over source
x_centre = x - x_source
y_centre = y - y_source

# Radial Distance from Centre
r = np.sqrt(x_centre**2 + y_centre**2)

# Avoid division by zero at the center
r[r==0] = 1e-6


# Calculate the angular component of the velocity
v_theta = 1 / r 
v_theta[r < R] = r[r<R]

# Convert to cartesian coordinates
u = v_theta * y_centre / r
v = -v_theta * x_centre / r

# Plot the vortex
plt.figure(figsize=(6, 6))
plt.quiver(x, y, u, v)

# Create circle
circle = plt.Circle((3,3), 1, color='red', fill=False)

# Add the circle to the plot
plt.gca().add_patch(circle)

# Set aspect ratio to be equal, so the circle looks circular
plt.gca().set_aspect('equal')


# Add source point
plt.scatter(x_source, y_source, color='red', marker='x', linewidths=2.5)

plt.title("Rankine Vortex, with Hurricane eye centered @ (3, 3) and Wall")
plt.show()
```

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Define alpha for 90 degree corner
alpha = 1.5 

# Create Meshgrid
x = np.arange(-10, 10, 0.5)
y = np.arange(-0.01, 10, 0.5)
X, Y = np.meshgrid(x, y)

# Calculate r, theta
r = np.sqrt(X**2 + Y**2)
theta = np.arctan2(Y, X)

# Avoid division by zero at the center
r[r==0] = 1e-6

# Calculate velocity components in polar coordinates
u_r = alpha * r**(alpha - 1) * np.cos(alpha * theta)
u_theta = -alpha * r**(alpha - 1) * np.sin(alpha * theta)

# Convert to cartesian coordinates
U = u_r * np.cos(theta) - u_theta * np.sin(theta)
V = u_r * np.sin(theta) + u_theta * np.cos(theta)
U = -U
V = -V

# Normalize the velocity components for better visualization
magnitude = np.sqrt(U**2 + V**2)
U /= magnitude
V /= magnitude

# Create the figure and the subplots
fig, ax = plt.subplots(figsize=(6, 6))

# Plot the corner flow as quiver
ax.quiver(X, Y, U, V, scale=20)

# Set aspect ratio to be equal
ax.set_aspect('equal')

ax.set_title("90 Degree Corner Flow")

# Show the plots
plt.tight_layout()
plt.show()
```
