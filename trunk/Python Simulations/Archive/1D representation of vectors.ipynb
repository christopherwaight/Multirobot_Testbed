```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt

# Define Constants
x_all = 40
x = np.linspace(0,31,x_all)

mag = np.array([0, 1, 2, 3, 3, 2, 1, 0, 0, 1, 2, 3, 3, 2, 1, 0, 0, 1, 2, 3, 3, 2, 1, 0, 0, 1, 2, 3, 3, 2, 1, 0,0, 1, 2, 3, 3, 2, 1, 0])
angle1 = np.array([0,0,0,0,np.pi/4,np.pi/4,np.pi/4,np.pi/4,np.pi/2,np.pi/2,np.pi/2,np.pi/2,0.75*np.pi,0.75*np.pi,0.75*np.pi,0.75*np.pi, np.pi, np.pi, np.pi, np.pi])
angle2 = -np.flip(angle1)
angle = np.concatenate((angle1, angle2))  # Combines angle1 and angle2 into one array


x = np.linspace(0, 255, x_all)

v = mag * np.sin(angle)
u = mag * np.cos(angle)

if len(u) != len(v):
    raise ValueError('Vectors u and v must be the same length.')

y = np.ones_like(x)  # Creates an array of 1's of the same size as x

# Plot a quiver plot
plt.quiver(x, y, u, v)

plt.show()
```
